# Example of calling PRIMAT from within python, and then reading the results which were stored in a .csv file.

```python
import subprocess
import os
import numpy as np
import csv
import matplotlib.pyplot as plt

# Set rcParams for LaTeX rendering
plt.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
    "font.serif": ["Computer Modern"],
    "lines.linewidth": 0.8,
    "figure.figsize": (6.8,5.2),
    #"xtick.labelsize": 8,
    #"ytick.labelsize": 8,
    #"axes.labelsize": 9,
    "ytick.right": True,
    "xtick.top": True,
    "ytick.direction": "in",
    "xtick.direction": "in",
})

currentdirectory = os.getcwd()
print(currentdirectory)

MathKernelCommand = '/Applications/Wolfram.app/Contents/MacOS/MathKernel'#Modify to match the location of Mathematcia Kernel
```

```python
filem = 'PyPRIMAT_FinalAbundances.m' #Input file which will load PRIMAT and dump the results in a csv file
desiredoutput = currentdirectory+"/Yi_final.csv" #Name of the csv file used for output
```

### Tools to prepare arguments which will be sent to Mathematica, and for calling PRIMAt inside Mathematica

```python
#This function converts a number in python/fortran format into a string that Mathematica can understand and convert to a number in Mathematica format
def stringfromnumeric(number):
    if isinstance(number,float):
        return "Interpreter[\"Number\"][StringDelete[\""+str(number)+"\",\" \"]]"
    else:
        return number
```

```python
def RunPRIMAT(DictArgs, filem, desiredoutput, MathKernelCommand):
    #This dictionary is packaged into a string which will be used as an argument to the Mathematica Kernel
    stringitems = [f"{key}={stringfromnumeric(value)}" for key, value in DictArgs.items()]

    ExtraArgumentsString = "; ".join(stringitems)+";"
    print(ExtraArgumentsString)
    
    #Outputfile is set to the desired output for Mathematica (as a string argument)
    optionoutput = "$Outputfile=\""+desiredoutput+"\""
    
    #Then we call the Mathematica script
    #We have two arguments to the Mathemtiaca script
    # 1) the ExtraArgumentsString which is a list of variables to update wrt the default ones before running the solver
    # 2) The name of the output file in which the script write a dictionary of the results
    return subprocess.call([MathKernelCommand,"-initfile",filem,ExtraArgumentsString,optionoutput])
```

### Basic example which runs PRIMAT and reads final abundances

```python
#Dictionary of parmameters and variables which needs to be overwritten wrt default parameters of PRIMAT.
DictArgs = {"$ReducedNetwork":True, 
            #"Meanh2\\[CapitalOmega]b0":0.022425,
            #"Mean\\[Tau]neutron":700.4,
            "Nrelat":0.,
            "$Verbose":False,
            #"$RecomputeWeakRates":True,
            }
```

```python
RunPRIMAT(DictArgs,filem,desiredoutput,MathKernelCommand)
```

```python
def ReadFinalresults(desiredoutput):
    with open(desiredoutput, mode='r', newline='') as csv_file:
        reader = csv.reader(csv_file)
        
        # Initialize an empty dictionary
        results = {}
        
        # Read each row and convert the values to float
        for row in reader:
            key = row[0]
            value = float(row[1])
            results[key] = value

    return results
```

```python
ReadFinalresults(desiredoutput)
```

## Using Parthenope rates for d+d reactions

```python
#Dictionary of parmameters and variables which needs to be overwritten wrt default parameters of PRIMAT.
DictArgs = {"$ReducedNetwork":True, 
            #"Meanh2\\[CapitalOmega]b0":0.022425,
            #"Mean\\[Tau]neutron":700.4,
            "Nrelat":0.,
            "$Parthenope":True,
            #"$RecomputeWeakRates":True,
            }

desiredoutputParthenope = currentdirectory+"/Yi_final__Parthenope.csv" #Name of the csv file used for output

RunPRIMAT(DictArgs,filem,desiredoutputParthenope,MathKernelCommand)

ReadFinalresults(desiredoutputParthenope)
```
